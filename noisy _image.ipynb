```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load a noisy image
image = cv2.imread('/content/sample_data/noise.jpeg', cv2.IMREAD_GRAYSCALE)

# Apply Box Filter
box_filtered = cv2.blur(image, (5, 5))  # Box filter with a 5x5 kernel

# Apply Gaussian Filter
gaussian_filtered = cv2.GaussianBlur(image, (5, 5), 1)  # Gaussian filter with a 5x5 kernel and sigma = 1

# Plot the results
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1), plt.imshow(image, cmap='gray'), plt.title('Original Noisy Image')
plt.subplot(1, 3, 2), plt.imshow(box_filtered, cmap='gray'), plt.title('Box Filtered')
plt.subplot(1, 3, 3), plt.imshow(gaussian_filtered, cmap='gray'), plt.title('Gaussian Filtered')
plt.show()
```
