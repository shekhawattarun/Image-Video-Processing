```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import math

# Load an image (grayscale for simplicity)
image = cv2.imread('/content/sample_data/cameraman.png', cv2.IMREAD_GRAYSCALE)

# Parameters
block_size = 4
height, width = image.shape

# Calculate number of blocks
num_blocks_x = width // block_size
num_blocks_y = height // block_size
num_blocks = num_blocks_x * num_blocks_y

# Determine subplot grid size
grid_size = math.ceil(math.sqrt(num_blocks))

# Create figure
fig, axes = plt.subplots(grid_size, grid_size, figsize=(10, 10))
axes = axes.flatten()

# Iterate over blocks
block_count = 0
for i in range(0, height, block_size):
    for j in range(0, width, block_size):
        if block_count < len(axes):
            # Extract 4x4 block
            block = image[i:i + block_size, j:j + block_size]

            # Display block in subplot
            axes[block_count].imshow(block, cmap='gray')
            axes[block_count].axis('off')
            axes[block_count].set_title(f'Block {block_count + 1}')
            block_count += 1

# Hide unused subplots
for k in range(block_count, len(axes)):
    axes[k].axis('off')

plt.tight_layout()
plt.show()
```
